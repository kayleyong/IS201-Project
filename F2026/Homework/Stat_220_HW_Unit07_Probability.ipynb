**Problem 2.** *by hand* A shop screens every build with an
automated flaky-test detector. Historically 4% of builds are genuinely flaky. The detector flags
90% of the genuinely flaky builds, and it also flags 15% of the perfectly good ones.

Part a. A build gets flagged. What is the probability it is genuinely flaky?

```python
p_flaky = 0.04
p_good = 1 - p_flaky
p_flag_given_flaky = 0.90
p_flag_given_good = 0.15

p_flaky_given_flag = (p_flag_given_flaky * p_flaky) / (
    (p_flag_given_flaky * p_flaky) + (p_flag_given_good * p_good)
)

print("P(flaky | flagged):", p_flaky_given_flag)
```

Part b. The vendor's brochure says "90% accurate." Write the two or three sentences you would say in a meeting to explain why that number does not answer the question anyone in the room is asking.

_Your answer:_ The vendor's 90% figure is the sensitivity, which only tells us how many genuinely flaky builds the detector catches, not whether a flagged build is actually flaky. What our team cares about is the reverse conditional, the predictive value, which asks how often a flagged build is a real problem. Because only 4% of our builds are genuinely flaky, the 15% false-flag rate on the remaining 96% of good builds overwhelms the true positives, meaning 80% of the flags we investigate will be false alarms.

Part c. An engineer can buy exactly one improvement: raise the 90% detection rate to 98%, or cut the 15% false flag rate to 10%. Which one most decreases the probability that a flagged build is a false positive?

```python
# Option 1: Raise detection rate to 98% (false flag rate stays 15%)
fp_opt1 = (0.15 * 0.96) / ((0.98 * 0.04) + (0.15 * 0.96))

# Option 2: Cut false flag rate to 10% (detection rate stays 90%)
fp_opt2 = (0.10 * 0.96) / ((0.90 * 0.04) + (0.10 * 0.96))

print("P(false positive | flagged) - Option 1 (98% detection):", fp_opt1)
print("P(false positive | flagged) - Option 2 (10% false flag):", fp_opt2)
```

_Your answer:_ Cutting the false flag rate from 15% to 10% decreases the false positive probability much more. Raising the detection rate to 98% only drops the chance that a flagged build is a false positive from 80% to about 78.6%, whereas cutting the false flag rate to 10% drops it to about 72.7%. Because 96% of builds are good, reducing errors on that much larger group has a far bigger impact on the share of false alarms.

**Problem 3.** A nightly cloud job costs compute credits

Part a. **By hand.** Compute $\mathbb{E}[X]$, then $\mathbb{E}[X^2]$, then $\mathrm{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2$, then $\mathrm{SD}(X)$.

```python
ex = (1 * 0.50) + (2 * 0.25) + (5 * 0.15) + (20 * 0.08) + (100 * 0.02)
ex2 = (1**2 * 0.50) + (2**2 * 0.25) + (5**2 * 0.15) + (20**2 * 0.08) + (100**2 * 0.02)
var_x = ex2 - ex**2
sd_x = var_x**0.5

print("E[X]:", ex)
print("E[X^2]:", ex2)
print("Var(X):", var_x)
print("SD(X):", sd_x)
```

Part b. **By hand.** Report the median of this distribution and compare it to the mean.

```python
cum_1 = 0.50
cum_2 = 0.50 + 0.25
cum_5 = 0.50 + 0.25 + 0.15
cum_20 = 0.50 + 0.25 + 0.15 + 0.08
cum_100 = 0.50 + 0.25 + 0.15 + 0.08 + 0.02

median = 1

print("P(X <= 1):", cum_1)
print("P(X <= 2):", cum_2)
print("P(X <= 5):", cum_5)
print("P(X <= 20):", cum_20)
print("P(X <= 100):", cum_100)
print("Median:", median)
print("Mean:", ex)
```

_Your answer:_ The median of this distribution is 1 credit, which is much lower than the mean of 5.35 credits. This large gap happens because the distribution is heavily right-skewed, with 50% of nights costing just 1 credit and 75% costing 2 credits or less. Even though a typical night is very cheap, the rare 20-credit and 100-credit nights in the upper tail pull the mean far above the median. As a result, the mean does not represent a typical night, and reporting both numbers together gives a much clearer picture of the cost distribution.

**Problem 6.** *Does the cap pay for itself?* The support plan from problem 4 costs 1 credit
a night and covers whatever a night's bill runs over 25. So what it saves you on one night is

$$W = \max(X - 25,\ 0)$$

which is 75 credits on a 100-credit night and 0 on every other night.

`norm` is already imported: `norm.cdf(z)` returns $\mathbb{P}(Z < z)$ for a standard normal.

Part a. Compute $\mathbb{E}[W]$ and $\mathrm{SD}(W)$ from the five outcomes of $X$.

```python
ew = (0 * 0.50) + (0 * 0.25) + (0 * 0.15) + (0 * 0.08) + (75 * 0.02)
ew2 = (0**2 * 0.50) + (0**2 * 0.25) + (0**2 * 0.15) + (0**2 * 0.08) + (75**2 * 0.02)
var_w = ew2 - ew**2
sd_w = var_w**0.5

print("E[W]:", ew)
print("SD(W):", sd_w)
```

Part b. Over 30 nights the plan costs 30 credits and saves a total of $S = W_1 + \cdots + W_{30}$. Give $\mathbb{E}[S]$ and $\mathrm{SD}(S)$, then use the central limit theorem to approximate $\mathbb{P}(S > 30)$, the chance the plan pays for itself over the month.

```python
es_30 = 30 * ew
var_s30 = 30 * var_w
sd_s30 = var_s30**0.5

z_30 = (30 - es_30) / sd_s30
p_gt_30 = 1 - norm.cdf(z_30)

print("E[S] (30 nights):", es_30)
print("SD(S) (30 nights):", sd_s30)
print("P(S > 30):", p_gt_30)
```

Part c. Do the same for 100 nights, where the plan costs 100 credits. Report $\mathbb{P}(S > 100)$ and say in one sentence why it is larger than your answer to part b, given that the plan saves 1.50 a night either way.

```python
es_100 = 100 * ew
var_s100 = 100 * var_w
sd_s100 = var_s100**0.5

z_100 = (100 - es_100) / sd_s100
p_gt_100 = 1 - norm.cdf(z_100)

print("E[S] (100 nights):", es_100)
print("SD(S) (100 nights):", sd_s100)
print("P(S > 100):", p_gt_100)
```

_Your answer:_ $P(S > 100)$ is larger than the 30-night probability because the expected profit grows proportionally with the number of nights ($n$), while the spread only grows with $\sqrt{n}$, making the plan more likely to come out ahead over more nights.

Part d. Now run the same approximation for a **single** night: treat $W$ as normal with the mean and standard deviation from part a, and approximate $\mathbb{P}(W > 1)$. Then compute the exact $\mathbb{P}(W > 1)$ straight from the table in problem 3. Report both.

```python
z_1 = (1 - ew) / sd_w
p_approx_1 = 1 - norm.cdf(z_1)
p_exact_1 = 0.02

print("Normal approximation P(W > 1):", p_approx_1)
print("Exact P(W > 1):", p_exact_1)
```

Part e. Your two numbers in part d are not close. Explain what went wrong, and state the condition the central limit theorem needs that one night does not meet.

_Your answer:_ The normal approximation gives about 51.9% while the exact probability is only 2% because a single night's savings is not bell-shaped at all. Instead, it is 0 on 98% of nights and 75 on the other 2%. The central limit theorem only applies to the sum or average of a large number of independent observations, so a single night does not meet the large sample size condition needed for the distribution to settle into a normal shape.
